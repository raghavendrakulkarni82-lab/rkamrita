```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    roc_auc_score
)
```

```python

import pandas as pd
import numpy as np
import re
import random

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score

# =====================================================
# 1. LOAD DATASET
# =====================================================

FILE_PATH = "/content/SDN_Intrusion.csv"

df = pd.read_csv(FILE_PATH, low_memory=False)

print("Original dataset shape:", df.shape)
print("Available classes:")
print(df["Class"].value_counts())

# Normalize label strings to handle encoding variations
def normalize_label(value):
    return re.sub(r"[^a-z0-9]+", " ",
                  str(value).lower()).strip()

normalized = df["Class"].apply(normalize_label)

# Identify the five traffic classes
class_patterns = {
    "BENIGN": r"\bbenign\b",
    "DDoS": r"\bddos\b",
    "Web Attack - Brute Force": r"brute force",
    "Web Attack - Sql Injection": r"sql injection",
    "Web Attack - XSS": r"\bxss\b"
}

# Select the required number of records from each class
quotas = {
    "BENIGN": 1100,
    "DDoS": 800,
    "Web Attack - Brute Force": 60,
    "Web Attack - Sql Injection": 60,
    "Web Attack - XSS": 18
}

selected_parts = []

for class_name, pattern in class_patterns.items():

    class_data = df[
        normalized.str.contains(pattern, regex=True, na=False)
    ]

    if len(class_data) < quotas[class_name]:
        raise ValueError(
            f"Not enough records for {class_name}. "
            f"Available: {len(class_data)}, "
            f"Required: {quotas[class_name]}"
        )

    selected_parts.append(
        class_data.sample(
            n=quotas[class_name],
            random_state=42
        )
    )

selected = pd.concat(selected_parts, ignore_index=True)
selected = selected.sample(frac=1, random_state=42).reset_index(drop=True)

print("\nSelected dataset shape:", selected.shape)
print("\nSelected class distribution:")
print(selected["Class"].apply(normalize_label).value_counts())

assert len(selected) == 2038

# =====================================================
# 2. SEPARATE FEATURES AND LABELS
# =====================================================

X = selected.drop(columns=["Class"]).copy()
y = selected["Class"].apply(normalize_label)

# Retain numerical features only
X = X.apply(pd.to_numeric, errors="coerce")

# Remove columns that contain no usable numerical values
X = X.dropna(axis=1, how="all")

# Remove infinite values
X = X.replace([np.inf, -np.inf], np.nan)

# Remove constant features
X = X.loc[:, X.nunique(dropna=True) > 1]

print("\nNumber of original numerical features:", X.shape[1])

# Encode five target classes
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)

print("\nClass encoding:")
for i, label in enumerate(label_encoder.classes_):
    print(i, ":", label)

# =====================================================
# 3. STRATIFIED TRAIN-TEST SPLIT
# =====================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.20,
    random_state=42,
    stratify=y_encoded
)

print("\nTraining samples:", len(X_train))
print("Testing samples:", len(X_test))

# =====================================================
# 4. MEAN IMPUTATION
# =====================================================

imputer = SimpleImputer(strategy="mean")

X_train_imp = imputer.fit_transform(X_train)
X_test_imp = imputer.transform(X_test)

# =====================================================
# 5. Z-SCORE NORMALIZATION
# =====================================================

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train_imp)
X_test_scaled = scaler.transform(X_test_imp)

print("\nMean imputation completed.")
print("Z-score normalization completed.")

# =====================================================
# 6. HAWO FEATURE SELECTION
# Hybrid ALO-inspired + WOA-inspired binary search
# =====================================================

class HAWO:

    def __init__(
        self,
        population_size=10,
        iterations=10,
        alpha=0.90,
        beta=0.10,
        random_state=42
    ):

        self.population_size = population_size
        self.iterations = iterations
        self.alpha = alpha
        self.beta = beta
        self.random_state = random_state

    def sigmoid(self, x):
        x = np.clip(x, -20, 20)
        return 1 / (1 + np.exp(-x))

    def evaluate(self, position, X_fit, y_fit, X_val, y_val):

        mask = position > 0.5

        # Ensure at least one feature is selected
        if not np.any(mask):
            mask[np.argmax(position)] = True

        model = RandomForestClassifier(
            n_estimators=80,
            max_depth=12,
            class_weight="balanced",
            random_state=self.random_state,
            n_jobs=-1
        )

        model.fit(X_fit[:, mask], y_fit)

        predictions = model.predict(X_val[:, mask])

        macro_f1 = f1_score(
            y_val,
            predictions,
            average="macro",
            zero_division=0
        )

        feature_ratio = np.sum(mask) / len(mask)

        # Minimize classification error and feature count
        fitness = (
            self.alpha * (1 - macro_f1)
            + self.beta * feature_ratio
        )

        return fitness

    def fit(self, X, y):

        rng = np.random.default_rng(self.random_state)

        # Internal validation split for feature selection
        X_fit, X_val, y_fit, y_val = train_test_split(
            X,
            y,
            test_size=0.20,
            random_state=self.random_state,
            stratify=y
        )

        n_features = X.shape[1]

        # Initialize continuous population
        population = rng.uniform(
            0.25, 0.75,
            size=(self.population_size, n_features)
        )

        best_position = None
        best_fitness = np.inf

        fitness_history = []

        for iteration in range(self.iterations):

            fitness_values = []

            for i in range(self.population_size):

                fitness = self.evaluate(
                    population[i],
                    X_fit, y_fit,
                    X_val, y_val
                )

                fitness_values.append(fitness)

                if fitness < best_fitness:
                    best_fitness = fitness
                    best_position = population[i].copy()

            fitness_values = np.array(fitness_values)

            # Best candidate acts as the current antlion/leader
            elite = population[np.argmin(fitness_values)].copy()

            new_population = population.copy()

            # Hybrid ALO-inspired and WOA-inspired updates
            for i in range(self.population_size):

                if i == np.argmin(fitness_values):
                    continue

                r = rng.random()
                A = 2 * (1 - iteration / self.iterations) * r - (
                    1 - iteration / self.iterations
                )

                C = 2 * rng.random()
                p = rng.random()

                if i % 2 == 0:
                    # ALO-inspired random-walk / elite update
                    random_walk = rng.normal(
                        0, 0.1, n_features
                    )

                    candidate = (
                        0.5 * population[i]
                        + 0.5 * elite
                        + random_walk
                    )

                else:
                    # WOA-inspired encircling / spiral update
                    if p < 0.5:

                        D = np.abs(
                            C * elite - population[i]
                        )

                        candidate = elite - A * D

                    else:

                        b = 1
                        l = rng.uniform(-1, 1)

                        D = np.abs(
                            elite - population[i]
                        )

                        candidate = (
                            D * np.exp(b * l)
                            * np.cos(2 * np.pi * l)
                            + elite
                        )

                # Convert continuous candidate into binary
                probabilities = self.sigmoid(candidate)

                new_population[i] = (
                    rng.random(n_features) < probabilities
                ).astype(float)

            population = np.clip(new_population, 0, 1)

            fitness_history.append(best_fitness)

            print(
                f"HAWO Iteration {iteration + 1}/{self.iterations}"
                f" | Best Fitness: {best_fitness:.5f}"
            )

        # Final selected feature mask
        self.selected_mask_ = best_position > 0.5

        if not np.any(self.selected_mask_):
            self.selected_mask_[np.argmax(best_position)] = True

        self.fitness_history_ = fitness_history

        return self

    def transform(self, X):
        return X[:, self.selected_mask_]

# =====================================================
# 7. APPLY HAWO
# =====================================================

hawо = HAWO(
    population_size=10,
    iterations=10,
    alpha=0.90,
    beta=0.10,
    random_state=42
)

hawо.fit(X_train_scaled, y_train)

X_train_hawo = hawо.transform(X_train_scaled)
X_test_hawo = hawо.transform(X_test_scaled)

# =====================================================
# 8. DISPLAY SELECTED FEATURES
# =====================================================

feature_names = X.columns.to_numpy()

selected_features = feature_names[hawо.selected_mask_]

print("\n========== HAWO FEATURE SELECTION ==========")

print("Original number of features:", len(feature_names))
print("Selected number of features:", len(selected_features))

print("\nSelected features:")
for feature in selected_features:
    print(feature)

reduction = (
    (len(feature_names) - len(selected_features))
    / len(feature_names)
) * 100

print("\nFeature reduction percentage:", round(reduction, 2), "%")

# =====================================================
# 9. SAVE PROCESSED FEATURE DATA
# =====================================================

processed_train = pd.DataFrame(
    X_train_hawo,
    columns=selected_features
)

processed_train["Class"] = y_train

processed_test = pd.DataFrame(
    X_test_hawo,
    columns=selected_features
)

processed_test["Class"] = y_test

processed_train.to_csv(
    "HAWO_processed_train.csv",
    index=False
)

processed_test.to_csv(
    "HAWO_processed_test.csv",
    index=False
)

print("\nProcessed training and testing files saved.")

print("\n========== FEATURE PROCESSING COMPLETED ==========")
```

```python

# ============================================================
# AERF: AUTOENCODER + RANDOM FOREST
# SDN MULTICLASS INTRUSION DETECTION
# ============================================================

import numpy as np
import pandas as pd
import tensorflow as tf
import joblib

from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_auc_score
)

# ============================================================
# 1. REPRODUCIBILITY
# ============================================================

SEED = 42

np.random.seed(SEED)
tf.keras.utils.set_random_seed(SEED)

# ============================================================
# 2. LOAD HAWO-PROCESSED DATA
# ============================================================

train_df = pd.read_csv("HAWO_processed_train.csv")
test_df = pd.read_csv("HAWO_processed_test.csv")

X_train = train_df.drop(columns=["Class"]).values.astype("float32")
X_test = test_df.drop(columns=["Class"]).values.astype("float32")

y_train_text = train_df["Class"].astype(str)
y_test_text = test_df["Class"].astype(str)

# ============================================================
# 3. ENCODE CLASS LABELS
# ============================================================

label_encoder = LabelEncoder()

y_train = label_encoder.fit_transform(y_train_text)

# Check that every test class was represented in training
unknown = set(y_test_text) - set(label_encoder.classes_)

if unknown:
    raise ValueError(
        f"Test set contains classes absent from training: {unknown}"
    )

y_test = label_encoder.transform(y_test_text)

num_classes = len(label_encoder.classes_)
input_dim = X_train.shape[1]

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
print("Selected input features:", input_dim)
print("Number of classes:", num_classes)
print("Classes:", label_encoder.classes_)

# ============================================================
# 4. BUILD AUTOENCODER
# ============================================================

input_layer = Input(shape=(input_dim,), name="input_features")

# Encoder
encoded = Dense(
    64,
    activation="relu",
    name="encoder_layer_1"
)(input_layer)

encoded = Dropout(0.10)(encoded)

latent = Dense(
    32,
    activation="relu",
    name="latent_representation"
)(encoded)

# Decoder
decoded = Dense(
    64,
    activation="relu",
    name="decoder_layer_1"
)(latent)

output_layer = Dense(
    input_dim,
    activation="linear",
    name="reconstructed_features"
)(decoded)

autoencoder = Model(
    inputs=input_layer,
    outputs=output_layer,
    name="AERF_Autoencoder"
)

autoencoder.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="mse"
)

print("\nAutoencoder architecture:")
autoencoder.summary()

# ============================================================
# 5. TRAIN AUTOENCODER
# ============================================================

early_stop = EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True
)

history = autoencoder.fit(
    X_train,
    X_train,
    epochs=50,
    batch_size=32,
    validation_split=0.20,
    callbacks=[early_stop],
    shuffle=True,
    verbose=1
)

# ============================================================
# 6. EXTRACT LATENT FEATURES
# ============================================================

encoder = Model(
    inputs=autoencoder.input,
    outputs=autoencoder.get_layer(
        "latent_representation"
    ).output,
    name="AERF_Encoder"
)

X_train_encoded = encoder.predict(
    X_train,
    verbose=0
)

X_test_encoded = encoder.predict(
    X_test,
    verbose=0
)

print("\nEncoded training shape:", X_train_encoded.shape)
print("Encoded testing shape:", X_test_encoded.shape)

# ============================================================
# 7. RANDOM FOREST CLASSIFIER
# ============================================================

rf_model = RandomForestClassifier(
    n_estimators=100,
    max_depth=None,
    min_samples_split=2,
    min_samples_leaf=1,
    class_weight="balanced",
    random_state=SEED,
    n_jobs=-1
)

rf_model.fit(
    X_train_encoded,
    y_train
)

# ============================================================
# 8. PREDICTION
# ============================================================

y_pred = rf_model.predict(X_test_encoded)

y_prob = rf_model.predict_proba(X_test_encoded)

# ============================================================
# 9. PERFORMANCE METRICS
# ============================================================

accuracy = accuracy_score(y_test, y_pred)

precision = precision_score(
    y_test,
    y_pred,
    average="macro",
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    average="macro",
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    average="macro",
    zero_division=0
)

cm = confusion_matrix(
    y_test,
    y_pred,
    labels=np.arange(num_classes)
)

# Multiclass macro one-vs-rest false-positive rate
fpr_values = []

for i in range(num_classes):

    TP = cm[i, i]
    FN = cm[i, :].sum() - TP
    FP = cm[:, i].sum() - TP
    TN = cm.sum() - TP - FN - FP

    fpr = FP / (FP + TN) if (FP + TN) > 0 else 0

    fpr_values.append(fpr)

macro_fpr = np.mean(fpr_values)

# Multiclass ROC-AUC
try:

    roc_auc = roc_auc_score(
        y_test,
        y_prob,
        multi_class="ovr",
        average="macro",
        labels=np.arange(num_classes)
    )

except ValueError:

    roc_auc = np.nan
    print("ROC-AUC could not be calculated.")

# ============================================================
# 10. DISPLAY RESULTS
# ============================================================

print("\n========================================")
print("       AERF DETECTION RESULTS")
print("========================================")

print(f"Accuracy:          {accuracy * 100:.2f}%")
print(f"Precision (Macro): {precision * 100:.2f}%")
print(f"Recall (Macro):    {recall * 100:.2f}%")
print(f"F1-score (Macro):  {f1 * 100:.2f}%")
print(f"False Positive Rate: {macro_fpr * 100:.2f}%")

if not np.isnan(roc_auc):
    print(f"ROC-AUC (Macro):   {roc_auc:.4f}")

print("\nConfusion Matrix:")
print(cm)

print("\nClassification Report:")

print(
    classification_report(
        y_test,
        y_pred,
        labels=np.arange(num_classes),
        target_names=label_encoder.classes_,
        zero_division=0
    )
)

# ============================================================
# 11. SAVE MODELS
# ============================================================

autoencoder.save("AERF_autoencoder.keras")

encoder.save("AERF_encoder.keras")

joblib.dump(
    rf_model,
    "AERF_random_forest.pkl"
)

joblib.dump(
    label_encoder,
    "AERF_label_encoder.pkl"
)

print("\nAERF models saved successfully.")

# ============================================================
# 12. FINAL VALIDATION
# ============================================================

print("\nTest samples:", len(y_test))
print("Confusion matrix total:", cm.sum())

assert cm.sum() == len(y_test)

print("\nAERF EVALUATION COMPLETED")
```

```python

import pandas as pd
import numpy as np
import random
import warnings

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score

warnings.filterwarnings("ignore")

# =====================================================
# STEP 1: LOAD DATASET
# =====================================================

df = pd.read_csv("SDN_Intrusion_1999.csv")

print("Original dataset shape:", df.shape)

# Separate input features and target
X = df.drop(columns=["Class"])
y = df["Class"].astype(str)

# Convert all feature columns to numeric
X = X.apply(pd.to_numeric, errors="coerce")

# Remove columns containing no usable values
X = X.dropna(axis=1, how="all")

# Remove constant columns
X = X.loc[:, X.nunique(dropna=True) > 1]

print("Usable feature count:", X.shape[1])

# Encode target labels
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)

print("\nClass mapping:")
for i, name in enumerate(label_encoder.classes_):
    print(i, "=", name)

# =====================================================
# STEP 2: STRATIFIED TRAIN-TEST SPLIT
# =====================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.20,
    random_state=42,
    stratify=y_encoded
)

print("\nTraining records:", len(X_train))
print("Testing records:", len(X_test))

# =====================================================
# STEP 3: MEAN IMPUTATION
# Fit only on training data
# =====================================================

imputer = SimpleImputer(strategy="mean")

X_train_imp = imputer.fit_transform(X_train)
X_test_imp = imputer.transform(X_test)

# =====================================================
# STEP 4: Z-SCORE NORMALIZATION
# Fit only on training data
# =====================================================

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train_imp)
X_test_scaled = scaler.transform(X_test_imp)

feature_names = np.array(X.columns)

print("\nMean imputation completed.")
print("Z-score normalization completed.")

# =====================================================
# STEP 5: HAWO FITNESS FUNCTION
# =====================================================

# Fitness combines classification error and feature
# subset size. Lower fitness is better.

ALPHA = 0.90
BETA = 0.10

def fitness_function(position):

    selected = position > 0.5

    # At least one feature must be selected
    if np.sum(selected) == 0:
        selected[np.argmax(position)] = True

    X_selected = X_train_scaled[:, selected]

    # Internal validation split from training data only
    X_tr, X_val, y_tr, y_val = train_test_split(
        X_selected,
        y_train,
        test_size=0.20,
        random_state=42,
        stratify=y_train
    )

    model = RandomForestClassifier(
        n_estimators=80,
        max_depth=12,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    )

    model.fit(X_tr, y_tr)

    predictions = model.predict(X_val)

    macro_f1 = f1_score(
        y_val,
        predictions,
        average="macro",
        zero_division=0
    )

    error = 1 - macro_f1
    feature_ratio = np.sum(selected) / len(selected)

    fitness = (
        ALPHA * error +
        BETA * feature_ratio
    )

    return fitness

# =====================================================
# STEP 6: HAWO OPTIMIZATION
# ALO-inspired and WOA-inspired updates
# =====================================================

POPULATION = 10
ITERATIONS = 10

n_features = X_train_scaled.shape[1]

np.random.seed(42)
random.seed(42)

population = np.random.rand(POPULATION, n_features)

fitness_values = np.array([
    fitness_function(individual)
    for individual in population
])

best_index = np.argmin(fitness_values)

best_position = population[best_index].copy()
best_fitness = fitness_values[best_index]

print("\nStarting HAWO optimization...")

for iteration in range(ITERATIONS):

    for i in range(POPULATION):

        current = population[i].copy()

        r = np.random.rand()
        A = 2 * r - 1
        C = 2 * np.random.rand()

        # ALO-inspired exploration
        if np.random.rand() < 0.5:

            antlion = best_position

            new_position = (
                current +
                np.random.rand(n_features) *
                (antlion - current)
            )

        # WOA-inspired exploitation
        else:

            D = np.abs(
                C * best_position - current
            )

            if np.random.rand() < 0.5:

                new_position = (
                    best_position - A * D
                )

            else:

                b = 1
                l = np.random.uniform(-1, 1)

                new_position = (
                    D * np.exp(b * l) *
                    np.cos(2 * np.pi * l) +
                    best_position
                )

        new_position = np.clip(
            new_position, 0, 1
        )

        new_fitness = fitness_function(new_position)

        if new_fitness < fitness_values[i]:

            population[i] = new_position
            fitness_values[i] = new_fitness

        if fitness_values[i] < best_fitness:

            best_fitness = fitness_values[i]
            best_position = population[i].copy()

    print(
        f"Iteration {iteration + 1}/{ITERATIONS} "
        f"- Best Fitness: {best_fitness:.5f}"
    )

# =====================================================
# STEP 7: SELECT OPTIMAL FEATURES
# =====================================================

selected_features = best_position > 0.5

if selected_features.sum() == 0:
    selected_features[np.argmax(best_position)] = True

selected_feature_names = feature_names[selected_features]

X_train_selected = X_train_scaled[:, selected_features]
X_test_selected = X_test_scaled[:, selected_features]

print("\n========== HAWO FEATURE SELECTION ==========")

print("Original features:", n_features)

print("Selected features:", len(selected_feature_names))

print("\nSelected feature names:")
print(selected_feature_names.tolist())

print("Best HAWO fitness:", best_fitness)

# =====================================================
# STEP 8: SAVE PROCESSED DATASETS
# =====================================================

train_df = pd.DataFrame(
    X_train_selected,
    columns=selected_feature_names
)

train_df["Class"] = y_train

test_df = pd.DataFrame(
    X_test_selected,
    columns=selected_feature_names
)

test_df["Class"] = y_test

train_df.to_csv(
    "HAWO_processed_train.csv",
    index=False
)

test_df.to_csv(
    "HAWO_processed_test.csv",
    index=False
)

# Save selected feature list
pd.DataFrame({
    "Selected_Features": selected_feature_names
}).to_csv(
    "HAWO_selected_features.csv",
    index=False
)

print("\n==========================================")
print("FEATURE PROCESSING COMPLETED")
print("==========================================")

print("Training dataset:", train_df.shape)
print("Testing dataset:", test_df.shape)

print("Saved files:")
print("1. HAWO_processed_train.csv")
print("2. HAWO_processed_test.csv")
print("3. HAWO_selected_features.csv")
```

```python

import numpy as np
import pandas as pd
import random

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer


# --------------------------------------------------
# 1. LOAD DATASET
# --------------------------------------------------

df = pd.read_csv("SDN_Intrusion_1999.csv")

# Separate features and target
X = df.drop(columns=["Class"])

# Convert textual class labels into numerical labels
class_mapping = {
    "BENIGN": 0,
    "DDoS": 1,
    "Web Attack - Brute Force": 2,
    "Web Attack - Sql Injection": 3,
    "Web Attack - XSS": 4
}

y = df["Class"].astype(str).str.strip().map(class_mapping)

# Check for unmapped class labels
if y.isnull().any():
    print("Unrecognized class labels:")
    print(df.loc[y.isnull(), "Class"].unique())
    raise ValueError("Some class labels are not included in class_mapping.")

y = y.astype(int)

print("Class mapping:")
for label, number in class_mapping.items():
    print(f"{number} = {label}")

print("\nClass distribution:")
print(y.value_counts().sort_index())

# Replace infinite values with missing values
X = X.replace([np.inf, -np.inf], np.nan)

# Remove columns that contain no usable information
X = X.dropna(axis=1, how="all")
X = X.loc[:, X.nunique(dropna=True) > 1]

# Stratified train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    stratify=y,
    random_state=42
)

# Fit preprocessing on training data only
imputer = SimpleImputer(strategy="mean")

X_train = pd.DataFrame(
    imputer.fit_transform(X_train),
    columns=X.columns
)

X_test = pd.DataFrame(
    imputer.transform(X_test),
    columns=X.columns
)

scaler = StandardScaler()

X_train = pd.DataFrame(
    scaler.fit_transform(X_train),
    columns=X.columns
)

X_test = pd.DataFrame(
    scaler.transform(X_test),
    columns=X.columns
)

# Internal validation split for optimization.
# The external test set remains untouched.
X_fit, X_val, y_fit, y_val = train_test_split(
    X_train, y_train,
    test_size=0.20,
    stratify=y_train,
    random_state=42
)


# --------------------------------------------------
# 2. HAWO PARAMETERS
# --------------------------------------------------

POPULATION_SIZE = 10
MAX_ITERATIONS = 10

ALPHA = 0.90       # Classification-error contribution
BETA = 0.10        # Feature-reduction contribution

N_FEATURES = X_train.shape[1]

random.seed(42)
np.random.seed(42)


# --------------------------------------------------
# 3. BINARY FEATURE REPRESENTATION
# --------------------------------------------------

def sigmoid(x):
    x = np.clip(x, -50, 50)
    return 1 / (1 + np.exp(-x))


def binary_position(position):
    probability = sigmoid(position)
    return (np.random.rand(len(position)) < probability).astype(int)


# --------------------------------------------------
# 4. FITNESS FUNCTION
# --------------------------------------------------

def fitness(position):

    selected = np.where(position == 1)[0]

    # At least one feature must be selected
    if len(selected) == 0:
        selected = [np.argmax(position)]

    Xtr = X_fit.iloc[:, selected]
    Xva = X_val.iloc[:, selected]

    model = RandomForestClassifier(
        n_estimators=80,
        max_depth=12,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    )

    model.fit(Xtr, y_fit)

    prediction = model.predict(Xva)

    macro_f1 = f1_score(
        y_val,
        prediction,
        average="macro",
        zero_division=0
    )

    feature_ratio = len(selected) / N_FEATURES

    # Lower fitness is better
    score = (
        ALPHA * (1 - macro_f1)
        + BETA * feature_ratio
    )

    return score


# --------------------------------------------------
# 5. INITIALIZE POPULATION
# --------------------------------------------------

population = np.random.uniform(
    -1, 1,
    (POPULATION_SIZE, N_FEATURES)
)

best_position = None
best_fitness = float("inf")

convergence_curve = []


# --------------------------------------------------
# 6. HAWO OPTIMIZATION
# --------------------------------------------------

for iteration in range(MAX_ITERATIONS):

    # Evaluate population
    binary_population = np.array([
        binary_position(individual)
        for individual in population
    ])

    fitness_values = np.array([
        fitness(individual)
        for individual in binary_population
    ])

    # Identify best candidate (Ant Lion / elite)
    best_index = np.argmin(fitness_values)

    if fitness_values[best_index] < best_fitness:
        best_fitness = fitness_values[best_index]
        best_position = population[best_index].copy()

    elite = best_position.copy()

    # ALO-inspired random-walk exploration
    # WOA-inspired encircling and spiral exploitation
    a = 2 - 2 * (iteration / MAX_ITERATIONS)

    for i in range(POPULATION_SIZE):

        current = population[i].copy()

        r1 = np.random.rand(N_FEATURES)
        r2 = np.random.rand(N_FEATURES)

        A = 2 * a * r1 - a
        C = 2 * r2

        p = np.random.rand()

        if p < 0.5:

            # WOA encircling mechanism
            D = np.abs(C * elite - current)

            new_position = elite - A * D

        else:

            # WOA spiral updating mechanism
            distance = np.abs(elite - current)

            l = np.random.uniform(-1, 1)

            b = 1

            new_position = (
                distance * np.exp(b * l)
                * np.cos(2 * np.pi * l)
                + elite
            )

        # ALO-inspired stochastic exploration
        if np.random.rand() < 0.30:

            random_walk = np.random.normal(
                0, 0.1, N_FEATURES
            )

            new_position += random_walk

        # Keep positions within bounds
        population[i] = np.clip(
            new_position, -5, 5
        )

    convergence_curve.append(best_fitness)

    print(
        f"Iteration {iteration + 1}/{MAX_ITERATIONS} "
        f"- Best Fitness: {best_fitness:.5f}"
    )


# --------------------------------------------------
# 7. OBTAIN OPTIMAL FEATURE SUBSET
# --------------------------------------------------

selected_binary = binary_position(best_position)

selected_indices = np.where(selected_binary == 1)[0]

if len(selected_indices) == 0:
    selected_indices = [np.argmax(best_position)]

selected_features = X_train.columns[selected_indices].tolist()

print("\n========== HAWO FEATURE SELECTION ==========")
print("Original features:", N_FEATURES)
print("Selected features:", len(selected_features))
print("Best fitness:", best_fitness)

print("\nSelected feature names:")
for feature in selected_features:
    print("-", feature)


# --------------------------------------------------
# 8. SAVE SELECTED DATA
# --------------------------------------------------

X_train_selected = X_train[selected_features].copy()
X_test_selected = X_test[selected_features].copy()

train_output = X_train_selected.copy()
train_output["Class"] = y_train.to_numpy()

test_output = X_test_selected.copy()
test_output["Class"] = y_test.to_numpy()

train_output.to_csv(
    "HAWO_processed_train.csv",
    index=False
)

test_output.to_csv(
    "HAWO_processed_test.csv",
    index=False
)

pd.DataFrame({
    "Selected_Feature": selected_features
}).to_csv(
    "HAWO_selected_features.csv",
    index=False
)

print("\nHAWO feature-selection files saved successfully.")
```

```python

# ============================================================
# ABLATION STUDY:
# 1. Random Forest
# 2. Autoencoder + Random Forest
# 3. HAWO + Autoencoder + Random Forest
# ============================================================

import numpy as np
import pandas as pd
import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, label_binarize
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    roc_auc_score
)

from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping


# ------------------------------------------------------------
# 1. REPRODUCIBILITY
# ------------------------------------------------------------

SEED = 42

np.random.seed(SEED)
tf.keras.utils.set_random_seed(SEED)


# ------------------------------------------------------------
# 2. LOAD DATASET AND ENCODE CLASS LABELS
# ------------------------------------------------------------
df = pd.read_csv("SDN_Intrusion_1999.csv")

# Remove leading and trailing spaces from dataset columns
df.columns = df.columns.str.strip()

class_mapping = {
    "BENIGN": 0,
    "DDoS": 1,
    "Web Attack - Brute Force": 2,
    "Web Attack - Sql Injection": 3,
    "Web Attack - XSS": 4
}

# ------------------------------------------------------------
# LOAD DATASET AND ENCODE CLASS LABELS
# ------------------------------------------------------------

import pandas as pd
import numpy as np

df = pd.read_csv("SDN_Intrusion_1999.csv")

# Clean column names
df.columns = df.columns.str.strip()

# Separate input features
X = df.drop(columns=["Class"])

# IMPORTANT: Define y_raw before using it
y_raw = df["Class"].astype(str).str.strip()

# Define class mapping
class_mapping = {
    "BENIGN": 0,
    "DDoS": 1,
    "Web Attack - Brute Force": 2,
    "Web Attack - Sql Injection": 3,
    "Web Attack - XSS": 4
}

# Normalize class names and map them to integers
normalized_mapping = {
    key.strip().lower(): value
    for key, value in class_mapping.items()
}

y = y_raw.str.lower().map(normalized_mapping)

# Check for unknown labels
if y.isnull().any():
    print("Unknown class labels:")
    print(y_raw[y.isnull()].unique())
    raise ValueError("Unknown class labels detected.")

y = y.astype(int)

# Display class information
print("Dataset shape:", df.shape)

print("\nClass mapping:")
for label, number in class_mapping.items():
    print(number, "=", label)

print("\nClass distribution:")
print(y.value_counts().sort_index())

if y.isnull().any():
    print("Unknown labels:", y_raw[y.isnull()].unique())
    raise ValueError("Unrecognized class labels found.")

y = y.astype(int)

class_names = [
    "BENIGN",
    "DDoS",
    "Web Attack - Brute Force",
    "Web Attack - Sql Injection",
    "Web Attack - XSS"
]

print("Dataset shape:", df.shape)
print("\nClass distribution:")
print(y.value_counts().sort_index())


# ------------------------------------------------------------
# 3. TRAIN-TEST SPLIT
# ------------------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    stratify=y,
    random_state=SEED
)

# Remove infinite values
X_train = X_train.replace([np.inf, -np.inf], np.nan)
X_test = X_test.replace([np.inf, -np.inf], np.nan)

# Remove entirely missing columns and constant columns
usable_columns = X_train.columns[
    X_train.notna().any(axis=0) &
    (X_train.nunique(dropna=True) > 1)
]

X_train = X_train[usable_columns]
X_test = X_test[usable_columns]

# Fit preprocessing on training data only
imputer = SimpleImputer(strategy="mean")

X_train = pd.DataFrame(
    imputer.fit_transform(X_train),
    columns=usable_columns,
    index=X_train.index
)

X_test = pd.DataFrame(
    imputer.transform(X_test),
    columns=usable_columns,
    index=X_test.index
)

scaler = StandardScaler()

X_train = pd.DataFrame(
    scaler.fit_transform(X_train),
    columns=usable_columns,
    index=X_train.index
)

X_test = pd.DataFrame(
    scaler.transform(X_test),
    columns=usable_columns,
    index=X_test.index
)

print("\nTraining samples:", len(X_train))
print("Testing samples:", len(X_test))
print("Available features:", X_train.shape[1])


# ------------------------------------------------------------
# 4. LOAD HAWO-SELECTED FEATURES
# ------------------------------------------------------------


# ------------------------------------------------------------
# 4. LOAD HAWO-SELECTED FEATURES (CORRECTED)
# ------------------------------------------------------------

selected_df = pd.read_csv("HAWO_selected_features.csv")

# Clean feature names in both files
selected_df["Selected_Feature"] = (
    selected_df["Selected_Feature"]
    .astype(str)
    .str.strip()
)

X_train.columns = X_train.columns.str.strip()
X_test.columns = X_test.columns.str.strip()

# Remove any accidental whitespace from selected feature names
selected_features = selected_df[
    "Selected_Feature"
].tolist()

# Remove duplicate feature names, preserving order
selected_features = list(dict.fromkeys(selected_features))

# Identify missing features
missing_features = [
    feature for feature in selected_features
    if feature not in X_train.columns
]

if missing_features:
    print("Features not found:", missing_features)

    raise ValueError(
        "Some HAWO-selected features do not match "
        "the original dataset columns."
    )

# Extract HAWO-selected features
X_train_hawo = X_train[selected_features].copy()
X_test_hawo = X_test[selected_features].copy()

print("\n========== HAWO FEATURE SELECTION ==========")
print("Original features:", X_train.shape[1])
print("Selected features:", len(selected_features))

print("\nSelected feature names:")
for feature in selected_features:
    print(feature)

print("\nHAWO training shape:", X_train_hawo.shape)
print("HAWO testing shape:", X_test_hawo.shape)


# ------------------------------------------------------------
# 5. AUTOENCODER FUNCTION
# ------------------------------------------------------------

def train_autoencoder(Xtr, Xte, model_name):

    input_dim = Xtr.shape[1]
    latent_dim = 4

    inputs = Input(shape=(input_dim,))

    encoded = Dense(64, activation="relu")(inputs)
    encoded = Dropout(0.1)(encoded)
    latent = Dense(latent_dim, activation="relu")(encoded)

    decoded = Dense(64, activation="relu")(latent)
    outputs = Dense(input_dim, activation="linear")(decoded)

    autoencoder = Model(inputs, outputs)

    encoder = Model(inputs, latent)

    autoencoder.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=0.001
        ),
        loss="mse"
    )

    early_stop = EarlyStopping(
        monitor="val_loss",
        patience=5,
        restore_best_weights=True
    )

    print(f"\nTraining autoencoder: {model_name}")

    autoencoder.fit(
        Xtr,
        Xtr,
        epochs=50,
        batch_size=32,
        validation_split=0.20,
        callbacks=[early_stop],
        verbose=0,
        shuffle=True
    )

    Xtr_encoded = encoder.predict(Xtr, verbose=0)
    Xte_encoded = encoder.predict(Xte, verbose=0)

    return Xtr_encoded, Xte_encoded


# ------------------------------------------------------------
# 6. RANDOM FOREST CLASSIFIER
# ------------------------------------------------------------

def train_random_forest(Xtr, ytr, Xte):

    rf = RandomForestClassifier(
        n_estimators=100,
        class_weight="balanced",
        random_state=SEED,
        n_jobs=-1
    )

    rf.fit(Xtr, ytr)

    predictions = rf.predict(Xte)
    probabilities = rf.predict_proba(Xte)

    return predictions, probabilities


# ------------------------------------------------------------
# 7. EVALUATION FUNCTION
# ------------------------------------------------------------

def evaluate_model(name, y_true, predictions, probabilities):

    accuracy = accuracy_score(y_true, predictions)

    precision = precision_score(
        y_true,
        predictions,
        average="macro",
        zero_division=0
    )

    recall = recall_score(
        y_true,
        predictions,
        average="macro",
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        predictions,
        average="macro",
        zero_division=0
    )

    cm = confusion_matrix(
        y_true,
        predictions,
        labels=[0, 1, 2, 3, 4]
    )

    # Macro one-vs-rest false-positive rate
    fpr_values = []

    for i in range(5):
        tp = cm[i, i]
        fn = cm[i, :].sum() - tp
        fp = cm[:, i].sum() - tp
        tn = cm.sum() - tp - fn - fp

        fpr = fp / (fp + tn) if (fp + tn) > 0 else 0
        fpr_values.append(fpr)

    macro_fpr = np.mean(fpr_values)

    y_binary = label_binarize(
        y_true,
        classes=[0, 1, 2, 3, 4]
    )

    try:
        auc = roc_auc_score(
            y_binary,
            probabilities,
            average="macro"
        )
    except ValueError:
        auc = np.nan

    results = {
        "Model": name,
        "Accuracy (%)": accuracy * 100,
        "Precision (%)": precision * 100,
        "Recall (%)": recall * 100,
        "F1-score (%)": f1 * 100,
        "FPR (%)": macro_fpr * 100,
        "ROC-AUC": auc
    }

    print("\n====================================")
    print("MODEL:", name)
    print("====================================")

    for metric, value in results.items():
        if metric != "Model":
            print(f"{metric}: {value:.4f}")

    print("\nConfusion Matrix:")
    print(cm)

    return results, cm


# ------------------------------------------------------------
# 8. EXPERIMENT 1: RANDOM FOREST ALONE
# ------------------------------------------------------------

print("\n\nEXPERIMENT 1: RANDOM FOREST")

rf_predictions, rf_probabilities = train_random_forest(
    X_train,
    y_train,
    X_test
)

rf_results, rf_cm = evaluate_model(
    "RF",
    y_test,
    rf_predictions,
    rf_probabilities
)


# ------------------------------------------------------------
# 9. EXPERIMENT 2: AUTOENCODER + RANDOM FOREST
#    Uses all available features
# ------------------------------------------------------------

print("\n\nEXPERIMENT 2: AUTOENCODER + RANDOM FOREST")

ae_train, ae_test = train_autoencoder(
    X_train,
    X_test,
    "AE-RF"
)

aerf_predictions, aerf_probabilities = train_random_forest(
    ae_train,
    y_train,
    ae_test
)

aerf_results, aerf_cm = evaluate_model(
    "AE-RF",
    y_test,
    aerf_predictions,
    aerf_probabilities
)


# ------------------------------------------------------------
# 10. EXPERIMENT 3: HAWO + AUTOENCODER + RANDOM FOREST
#     Uses HAWO-selected features
# ------------------------------------------------------------

print("\n\nEXPERIMENT 3: HAWO + AE + RANDOM FOREST")

hawo_ae_train, hawo_ae_test = train_autoencoder(
    X_train_hawo,
    X_test_hawo,
    "HAWO-AE-RF"
)

hawo_predictions, hawo_probabilities = train_random_forest(
    hawo_ae_train,
    y_train,
    hawo_ae_test
)

hawo_results, hawo_cm = evaluate_model(
    "HAWO-AE-RF",
    y_test,
    hawo_predictions,
    hawo_probabilities
)


# ------------------------------------------------------------
# 11. FINAL ABLATION COMPARISON TABLE
# ------------------------------------------------------------

comparison = pd.DataFrame([
    rf_results,
    aerf_results,
    hawo_results
])

print("\n\n========== ABLATION STUDY RESULTS ==========")

print(
    comparison.round(4).to_string(index=False)
)

comparison.to_csv(
    "AERF_HAWO_Ablation_Results.csv",
    index=False
)

print("\nResults saved as AERF_HAWO_Ablation_Results.csv")


# ------------------------------------------------------------
# 12. SAVE CONFUSION MATRICES
# ------------------------------------------------------------

for name, matrix in [
    ("RF", rf_cm),
    ("AE_RF", aerf_cm),
    ("HAWO_AE_RF", hawo_cm)
]:

    matrix_df = pd.DataFrame(
        matrix,
        index=class_names,
        columns=class_names
    )

    matrix_df.to_csv(
        f"{name}_Confusion_Matrix.csv"
    )

print("\nAll confusion matrices saved successfully.")
```

```python

# ============================================================
# STATISTICAL VARIANCE + ABLATION STUDY
# RF vs AE-RF vs HAWO + AE-RF
# 10 independent experimental runs
# ============================================================

import os
import random
import warnings
import numpy as np
import pandas as pd
import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, label_binarize
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    roc_auc_score
)

from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings("ignore")

# ------------------------------------------------------------
# 1. FILE PATHS
# ------------------------------------------------------------

DATA_FILE = "SDN_Intrusion_1999.csv"

# This file must contain the feature names selected by HAWO.
HAWO_FILE = "HAWO_selected_features.csv"

OUTPUT_DIR = "Statistical_Results"
os.makedirs(OUTPUT_DIR, exist_ok=True)

# ------------------------------------------------------------
# 2. EXPERIMENT SETTINGS
# ------------------------------------------------------------

N_RUNS = 10
TEST_SIZE = 0.20

RF_TREES = 100
LATENT_DIM = 4
AE_EPOCHS = 50
AE_BATCH_SIZE = 32

RANDOM_SEEDS = [42, 52, 62, 72, 82,
                92, 102, 112, 122, 132]

# ------------------------------------------------------------
# 3. LOAD DATASET
# ------------------------------------------------------------

df = pd.read_csv(DATA_FILE)

# Remove accidental spaces from column names
df.columns = df.columns.astype(str).str.strip()

print("Dataset shape:", df.shape)

if "Class" not in df.columns:
    raise ValueError(
        "Class column not found. Check the dataset column names."
    )

# ------------------------------------------------------------
# 4. ENCODE CLASS LABELS
# ------------------------------------------------------------

class_mapping = {
    "BENIGN": 0,
    "DDoS": 1,
    "Web Attack - Brute Force": 2,
    "Web Attack - Sql Injection": 3,
    "Web Attack - XSS": 4
}

y_raw = df["Class"].astype(str).str.strip()

normalized_mapping = {
    key.lower(): value
    for key, value in class_mapping.items()
}

y = y_raw.str.lower().map(normalized_mapping)

if y.isnull().any():
    unknown = sorted(y_raw[y.isnull()].unique())
    raise ValueError(
        f"Unknown class labels found: {unknown}"
    )

y = y.astype(int)

# ------------------------------------------------------------
# 5. PREPARE FEATURES
# ------------------------------------------------------------

X = df.drop(columns=["Class"]).copy()

# Convert columns to numeric
X = X.apply(pd.to_numeric, errors="coerce")

# Remove columns containing no valid values
X = X.dropna(axis=1, how="all")

# Remove constant columns
X = X.loc[:, X.nunique(dropna=True) > 1]

# Replace infinite values
X = X.replace([np.inf, -np.inf], np.nan)

all_features = X.columns.tolist()

print("Usable features:", len(all_features))
print("\nClass distribution:")
print(y.value_counts().sort_index())

# ------------------------------------------------------------
# 6. LOAD HAWO SELECTED FEATURES
# ------------------------------------------------------------

selected_df = pd.read_csv(HAWO_FILE)

selected_df.columns = (
    selected_df.columns.astype(str).str.strip()
)

if "Selected_Feature" not in selected_df.columns:
    raise ValueError(
        "HAWO file must contain a 'Selected_Feature' column."
    )

hawо_features = (
    selected_df["Selected_Feature"]
    .dropna()
    .astype(str)
    .str.strip()
    .tolist()
)

# Remove duplicate feature names, preserving order
hawо_features = list(dict.fromkeys(hawо_features))

missing_features = [
    f for f in hawо_features if f not in all_features
]

if missing_features:
    raise ValueError(
        "HAWO features do not match dataset columns:\n"
        + "\n".join(missing_features)
        + "\nCheck whether the HAWO file belongs to this dataset."
    )

if not hawо_features:
    raise ValueError("No HAWO-selected features were found.")

print("\nHAWO-selected feature count:", len(hawо_features))

# ------------------------------------------------------------
# 7. EVALUATION FUNCTION
# ------------------------------------------------------------

def evaluate_model(model_name, y_true, y_pred, y_prob):

    cm = confusion_matrix(
        y_true,
        y_pred,
        labels=[0, 1, 2, 3, 4]
    )

    # Macro one-vs-rest FPR
    fpr_values = []

    for i in range(5):
        TP = cm[i, i]
        FN = cm[i, :].sum() - TP
        FP = cm[:, i].sum() - TP
        TN = cm.sum() - TP - FN - FP

        fpr = FP / (FP + TN) if (FP + TN) > 0 else 0
        fpr_values.append(fpr)

    macro_fpr = np.mean(fpr_values)

    # Multiclass one-vs-rest ROC-AUC
    y_bin = label_binarize(y_true, classes=[0, 1, 2, 3, 4])

    try:
        auc = roc_auc_score(
            y_bin,
            y_prob,
            average="macro",
            multi_class="ovr"
        )
    except ValueError:
        auc = np.nan

    return {
        "Model": model_name,
        "Accuracy": accuracy_score(y_true, y_pred),
        "Macro_Precision": precision_score(
            y_true, y_pred,
            average="macro",
            zero_division=0
        ),
        "Macro_Recall": recall_score(
            y_true, y_pred,
            average="macro",
            zero_division=0
        ),
        "Macro_F1": f1_score(
            y_true, y_pred,
            average="macro",
            zero_division=0
        ),
        "Macro_FPR": macro_fpr,
        "Macro_ROC_AUC": auc
    }, cm

# ------------------------------------------------------------
# 8. AUTOENCODER + RANDOM FOREST
# ------------------------------------------------------------

def train_ae_rf(X_train, X_test, y_train, seed):

    tf.keras.backend.clear_session()
    tf.keras.utils.set_random_seed(seed)

    input_dim = X_train.shape[1]

    if input_dim < 2:
        raise ValueError("At least two input features are required.")

    input_layer = Input(shape=(input_dim,))

    encoded = Dense(64, activation="relu")(input_layer)
    encoded = Dropout(0.1)(encoded)

    latent = Dense(
        LATENT_DIM,
        activation="relu",
        name="latent"
    )(encoded)

    decoded = Dense(64, activation="relu")(latent)

    output_layer = Dense(
        input_dim,
        activation="linear"
    )(decoded)

    autoencoder = Model(input_layer, output_layer)

    autoencoder.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
        loss="mse"
    )

    early_stop = EarlyStopping(
        monitor="val_loss",
        patience=5,
        restore_best_weights=True
    )

    autoencoder.fit(
        X_train,
        X_train,
        epochs=AE_EPOCHS,
        batch_size=AE_BATCH_SIZE,
        validation_split=0.20,
        callbacks=[early_stop],
        verbose=0,
        shuffle=True
    )

    encoder = Model(
        inputs=autoencoder.input,
        outputs=autoencoder.get_layer("latent").output
    )

    Z_train = encoder.predict(X_train, verbose=0)
    Z_test = encoder.predict(X_test, verbose=0)

    rf = RandomForestClassifier(
        n_estimators=RF_TREES,
        max_depth=None,
        class_weight="balanced",
        random_state=seed,
        n_jobs=-1
    )

    rf.fit(Z_train, y_train)

    y_pred = rf.predict(Z_test)
    y_prob = rf.predict_proba(Z_test)

    # Ensure probability columns align with class labels 0–4
    probabilities = np.zeros((len(X_test), 5))
    for j, cls in enumerate(rf.classes_):
        probabilities[:, int(cls)] = y_prob[:, j]

    return y_pred, probabilities

# ------------------------------------------------------------
# 9. RUN ALL EXPERIMENTS
# ------------------------------------------------------------

all_results = []
all_confusion_matrices = []

for run_number, seed in enumerate(RANDOM_SEEDS, start=1):

    print(f"\n========== RUN {run_number}/{N_RUNS} ==========")
    print("Random seed:", seed)

    # Set random seeds
    random.seed(seed)
    np.random.seed(seed)
    tf.keras.utils.set_random_seed(seed)

    # Stratified train-test split
    X_train_raw, X_test_raw, y_train, y_test = train_test_split(
        X,
        y,
        test_size=TEST_SIZE,
        random_state=seed,
        stratify=y
    )

    # Fit preprocessing ONLY on training data
    imputer = SimpleImputer(strategy="median")

    X_train_imp = imputer.fit_transform(X_train_raw)
    X_test_imp = imputer.transform(X_test_raw)

    scaler = StandardScaler()

    X_train_scaled = scaler.fit_transform(X_train_imp)
    X_test_scaled = scaler.transform(X_test_imp)

    X_train_all = pd.DataFrame(
        X_train_scaled,
        columns=all_features,
        index=X_train_raw.index
    )

    X_test_all = pd.DataFrame(
        X_test_scaled,
        columns=all_features,
        index=X_test_raw.index
    )

    # --------------------------------------------------------
    # MODEL 1: RANDOM FOREST WITH ALL FEATURES
    # --------------------------------------------------------

    rf = RandomForestClassifier(
        n_estimators=RF_TREES,
        max_depth=None,
        class_weight="balanced",
        random_state=seed,
        n_jobs=-1
    )

    rf.fit(X_train_all, y_train)

    pred_rf = rf.predict(X_test_all)
    prob_rf_raw = rf.predict_proba(X_test_all)

    prob_rf = np.zeros((len(X_test_all), 5))

    for j, cls in enumerate(rf.classes_):
        prob_rf[:, int(cls)] = prob_rf_raw[:, j]

    result, cm = evaluate_model(
        "RF",
        y_test,
        pred_rf,
        prob_rf
    )

    result["Run"] = run_number
    result["Seed"] = seed
    result["Selected_Features"] = len(all_features)

    all_results.append(result)

    all_confusion_matrices.append({
        "Run": run_number,
        "Model": "RF",
        "Confusion_Matrix": cm.tolist()
    })

    print("RF Accuracy:", round(result["Accuracy"], 4))

    # --------------------------------------------------------
    # MODEL 2: AE-RF WITH ALL FEATURES
    # --------------------------------------------------------

    pred_ae, prob_ae = train_ae_rf(
        X_train_all.values,
        X_test_all.values,
        y_train,
        seed
    )

    result, cm = evaluate_model(
        "AE-RF",
        y_test,
        pred_ae,
        prob_ae
    )

    result["Run"] = run_number
    result["Seed"] = seed
    result["Selected_Features"] = len(all_features)

    all_results.append(result)

    all_confusion_matrices.append({
        "Run": run_number,
        "Model": "AE-RF",
        "Confusion_Matrix": cm.tolist()
    })

    print("AE-RF Accuracy:", round(result["Accuracy"], 4))

    # --------------------------------------------------------
    # MODEL 3: HAWO + AE-RF
    # --------------------------------------------------------

    X_train_hawo = X_train_all[hawо_features]
    X_test_hawo = X_test_all[hawо_features]

    pred_hawo, prob_hawo = train_ae_rf(
        X_train_hawo.values,
        X_test_hawo.values,
        y_train,
        seed
    )

    result, cm = evaluate_model(
        "HAWO-AE-RF",
        y_test,
        pred_hawo,
        prob_hawo
    )

    result["Run"] = run_number
    result["Seed"] = seed
    result["Selected_Features"] = len(hawо_features)

    all_results.append(result)

    all_confusion_matrices.append({
        "Run": run_number,
        "Model": "HAWO-AE-RF",
        "Confusion_Matrix": cm.tolist()
    })

    print("HAWO-AE-RF Accuracy:", round(result["Accuracy"], 4))

# ------------------------------------------------------------
# 10. SAVE INDIVIDUAL RUN RESULTS
# ------------------------------------------------------------

results_df = pd.DataFrame(all_results)

results_path = os.path.join(
    OUTPUT_DIR,
    "Individual_Run_Results.csv"
)

results_df.to_csv(results_path, index=False)

# ------------------------------------------------------------
# 11. CALCULATE MEAN AND STANDARD DEVIATION
# ------------------------------------------------------------

metric_columns = [
    "Accuracy",
    "Macro_Precision",
    "Macro_Recall",
    "Macro_F1",
    "Macro_FPR",
    "Macro_ROC_AUC"
]

summary_rows = []

for model_name in results_df["Model"].unique():

    model_data = results_df[
        results_df["Model"] == model_name
    ]

    for metric in metric_columns:

        values = model_data[metric].dropna()

        summary_rows.append({
            "Model": model_name,
            "Metric": metric,
            "Mean": values.mean(),
            "Standard_Deviation": values.std(ddof=1),
            "Mean_Percent": values.mean() * 100,
            "SD_Percent": values.std(ddof=1) * 100,
            "Runs": len(values)
        })

summary_df = pd.DataFrame(summary_rows)

summary_path = os.path.join(
    OUTPUT_DIR,
    "Mean_SD_Statistical_Results.csv"
)

summary_df.to_csv(summary_path, index=False)

# ------------------------------------------------------------
# 12. SAVE CONFUSION MATRICES
# ------------------------------------------------------------

cm_df = pd.DataFrame(all_confusion_matrices)

cm_path = os.path.join(
    OUTPUT_DIR,
    "Confusion_Matrices_All_Runs.csv"
)

cm_df.to_csv(cm_path, index=False)

# ------------------------------------------------------------
# 13. DISPLAY RESULTS
# ------------------------------------------------------------

print("\n\n======================================")
print("INDIVIDUAL EXPERIMENTAL RESULTS")
print("======================================")

print(results_df.round(4).to_string(index=False))

print("\n\n======================================")
print("MEAN AND STANDARD DEVIATION")
print("======================================")

print(summary_df.round(4).to_string(index=False))

print("\n\nFiles saved in:", OUTPUT_DIR)
print("1. Individual_Run_Results.csv")
print("2. Mean_SD_Statistical_Results.csv")
print("3. Confusion_Matrices_All_Runs.csv")
```
